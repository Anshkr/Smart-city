```python
# STEP 2: Cities
cities = {
    "Delhi": (28.6139, 77.2090),
    "Mumbai": (19.0760, 72.8777),
    "Bengaluru": (12.9716, 77.5946),
    "Chennai": (13.0827, 80.2707),
    "Kolkata": (22.5726, 88.3639),
    "Hyderabad": (17.3850, 78.4867)
}
```

```python

# STEP 3: Connect to Open-Meteo Air Quality API

url = "https://air-quality-api.open-meteo.com/v1/air-quality"

all_records = []

for city, (lat, lon) in cities.items():

    params = {
        "latitude": lat,
        "longitude": lon,
        "hourly": "pm10,pm2_5,nitrogen_dioxide,sulphur_dioxide,carbon_monoxide,ozone",
        "past_days": 1,
        "forecast_days": 1
    }

    response = requests.get(url, params=params, timeout=30)
    response.raise_for_status()

    data = response.json()
    hourly = data["hourly"]

# Convert hourly API data into individual records

    for i, timestamp in enumerate(hourly["time"]):

        record = {
            "city": city,
            "timestamp": timestamp,
            "pm10": hourly["pm10"][i],
            "pm2_5": hourly["pm2_5"][i],
            "nitrogen_dioxide": hourly["nitrogen_dioxide"][i],
            "sulphur_dioxide": hourly["sulphur_dioxide"][i],
            "carbon_monoxide": hourly["carbon_monoxide"][i],
            "ozone": hourly["ozone"][i],
            "source": "Open-Meteo"
        }

        all_records.append(record)

    print(f"{city}: Extraction successful")
```

```python
# Convert extracted records into a Pandas DataFrame
air_quality_df = pd.DataFrame(all_records)

print("\nTotal records extracted:", len(air_quality_df))

# Display extracted data
display(air_quality_df)
```

```python

# STEP 5: Basic data verification

print("Total rows:", len(air_quality_df))

print("\nColumns:")
print(air_quality_df.columns.tolist())

print("\nRecords per city:")
print(air_quality_df.groupby("city").size())

print("\nMissing values:")
display(air_quality_df.isnull().sum().to_frame("Missing Values"))

print("\nSample records:")
display(air_quality_df.head(10))
```

```python

## Forecast
# STEP 6: Extract weather data for six cities

import requests
import pandas as pd

weather_url = "https://api.open-meteo.com/v1/forecast"

weather_records = []

for city, (lat, lon) in cities.items():

    params = {
        "latitude": lat,
        "longitude": lon,
        "hourly": "temperature_2m,relative_humidity_2m,wind_speed_10m,precipitation",
        "past_days": 1,
        "forecast_days": 1,
        "timezone": "Asia/Kolkata"
    }

    response = requests.get(
        weather_url,
        params=params,
        timeout=30
    )

    response.raise_for_status()

    data = response.json()
    hourly = data["hourly"]

    for i, timestamp in enumerate(hourly["time"]):

        record = {
            "city": city,
            "timestamp": timestamp,
            "temperature": hourly["temperature_2m"][i],
            "humidity": hourly["relative_humidity_2m"][i],
            "wind_speed": hourly["wind_speed_10m"][i],
            "precipitation": hourly["precipitation"][i],
            "source": "Open-Meteo"
        }

        weather_records.append(record)

    print(f"{city}: Weather extraction successful")

# Convert records into DataFrame
weather_df = pd.DataFrame(weather_records)

print("\nTotal weather records:", len(weather_df))

display(weather_df.head(10))
```

```python

# STEP 9: Extract NASA POWER data for all six cities

nasa_records = []

for city, (lat, lon) in cities.items():

    params = {
        "parameters": "T2M,T2M_MAX,T2M_MIN,RH2M,WS10M,PRECTOTCORR",
        "community": "RE",
        "longitude": lon,
        "latitude": lat,
        "start": start_date.strftime("%Y%m%d"),
        "end": end_date.strftime("%Y%m%d"),
        "format": "JSON",
        "time-standard": "UTC"
    }

    try:
        response = requests.get(
            nasa_url,
            params=params,
            timeout=30
        )

        response.raise_for_status()

        data = response.json()
        parameters_data = data["properties"]["parameter"]

        for timestamp in parameters_data["T2M"].keys():

            def clean_value(value):
                if value is None or float(value) <= -900:
                    return None
                return float(value)

            nasa_records.append({
                "city": city,
                "date": timestamp,
                "temperature_avg": clean_value(
                    parameters_data["T2M"].get(timestamp)
                ),
                "temperature_max": clean_value(
                    parameters_data["T2M_MAX"].get(timestamp)
                ),
                "temperature_min": clean_value(
                    parameters_data["T2M_MIN"].get(timestamp)
                ),
                "humidity": clean_value(
                    parameters_data["RH2M"].get(timestamp)
                ),
                "wind_speed": clean_value(
                    parameters_data["WS10M"].get(timestamp)
                ),
                "precipitation": clean_value(
                    parameters_data["PRECTOTCORR"].get(timestamp)
                ),
                "source": "NASA POWER"
            })

        print(f"{city}: NASA extraction successful")

    except Exception as e:
        print(f"{city}: Extraction failed - {e}")

nasa_df = pd.DataFrame(nasa_records)

print("Total NASA records:", len(nasa_df))

display(nasa_df.head(10))
```

```python

# STEP 11: Align all three datasets to the same dates

import requests
import pandas as pd

start_str = start_date.strftime("%Y-%m-%d")
end_str = end_date.strftime("%Y-%m-%d")

print("Common date range:", start_str, "to", end_str)

# ---------------------------------
# 1. AIR QUALITY DATA
# ---------------------------------

air_url = "https://air-quality-api.open-meteo.com/v1/air-quality"

air_records_aligned = []

for city, (lat, lon) in cities.items():

    params = {
        "latitude": lat,
        "longitude": lon,
        "hourly": "pm10,pm2_5,nitrogen_dioxide,sulphur_dioxide,carbon_monoxide,ozone",
        "past_days": 10,
        "forecast_days": 0,
        "timezone": "UTC"
    }

    response = requests.get(air_url, params=params, timeout=30)
    response.raise_for_status()

    hourly = response.json()["hourly"]

    for i, timestamp in enumerate(hourly["time"]):

        record_date = timestamp[:10]

        if start_str <= record_date <= end_str:
            air_records_aligned.append({
                "city": city,
                "timestamp": timestamp,
                "pm10": hourly["pm10"][i],
                "pm2_5": hourly["pm2_5"][i],
                "nitrogen_dioxide": hourly["nitrogen_dioxide"][i],
                "sulphur_dioxide": hourly["sulphur_dioxide"][i],
                "carbon_monoxide": hourly["carbon_monoxide"][i],
                "ozone": hourly["ozone"][i],
                "source": "Open-Meteo Air Quality"
            })

    print(city, "Air Quality completed")

air_quality_df = pd.DataFrame(air_records_aligned)

# ---------------------------------
# 2. WEATHER DATA
# ---------------------------------

weather_url = "https://api.open-meteo.com/v1/forecast"

weather_records_aligned = []

for city, (lat, lon) in cities.items():

    params = {
        "latitude": lat,
        "longitude": lon,
        "hourly": "temperature_2m,relative_humidity_2m,wind_speed_10m,precipitation",
        "past_days": 10,
        "forecast_days": 0,
        "timezone": "UTC"
    }

    response = requests.get(weather_url, params=params, timeout=30)
    response.raise_for_status()

    hourly = response.json()["hourly"]

    for i, timestamp in enumerate(hourly["time"]):

        record_date = timestamp[:10]

        if start_str <= record_date <= end_str:
            weather_records_aligned.append({
                "city": city,
                "timestamp": timestamp,
                "temperature": hourly["temperature_2m"][i],
                "humidity": hourly["relative_humidity_2m"][i],
                "wind_speed": hourly["wind_speed_10m"][i],
                "precipitation": hourly["precipitation"][i],
                "source": "Open-Meteo Weather"
            })

    print(city, "Weather completed")

weather_df = pd.DataFrame(weather_records_aligned)

# ---------------------------------
# 3. CHECK ALL THREE DATASETS
# ---------------------------------

print("\nAIR QUALITY:", len(air_quality_df))
print("WEATHER:", len(weather_df))
print("NASA POWER:", len(nasa_df))

print("\nAir Quality date range:")
print(air_quality_df["timestamp"].min(),
      air_quality_df["timestamp"].max())

print("\nWeather date range:")
print(weather_df["timestamp"].min(),
      weather_df["timestamp"].max())

print("\nNASA sample:")
display(nasa_df.head())

print("\nAir Quality sample:")
display(air_quality_df.head())

print("\nWeather sample:")
display(weather_df.head())
```

```python

import pandas as pd

# -----------------------------------------
# STEP 1: CHECK INPUT DATASETS
# -----------------------------------------

print("Air Quality Records:", len(air_quality_df))
print("Weather Records:", len(weather_df))
print("NASA Records:", len(nasa_df))

# -----------------------------------------
# STEP 2: CONVERT HOURLY TIMESTAMPS TO DATES
# -----------------------------------------

air_quality_df = air_quality_df.copy()
weather_df = weather_df.copy()
nasa_df = nasa_df.copy()

air_quality_df["timestamp"] = pd.to_datetime(
    air_quality_df["timestamp"], utc=True
)

weather_df["timestamp"] = pd.to_datetime(
    weather_df["timestamp"], utc=True
)

air_quality_df["date"] = (
    air_quality_df["timestamp"].dt.strftime("%Y-%m-%d")
)

weather_df["date"] = (
    weather_df["timestamp"].dt.strftime("%Y-%m-%d")
)

nasa_df["date"] = pd.to_datetime(
    nasa_df["date"]
).dt.strftime("%Y-%m-%d")

# -----------------------------------------
# STEP 3: AGGREGATE AIR QUALITY DATA
# Calculate daily average pollution levels
# -----------------------------------------

pollutants = [
    "pm10",
    "pm2_5",
    "nitrogen_dioxide",
    "sulphur_dioxide",
    "carbon_monoxide",
    "ozone"
]

air_daily = (
    air_quality_df
    .groupby(["city", "date"])[pollutants]
    .mean()
    .reset_index()
)

# -----------------------------------------
# STEP 4: AGGREGATE WEATHER DATA
# Average weather measurements
# Sum daily precipitation
# -----------------------------------------

weather_daily = (
    weather_df
    .groupby(["city", "date"])
    .agg(
        weather_temperature_avg=("temperature", "mean"),
        weather_humidity_avg=("humidity", "mean"),
        weather_wind_speed_avg=("wind_speed", "mean"),
        weather_precipitation_sum=("precipitation", "sum")
    )
    .reset_index()
)

# -----------------------------------------
# STEP 5: PREPARE NASA DATA
# Rename columns to avoid confusion
# -----------------------------------------

nasa_daily = nasa_df.rename(columns={
    "temperature_avg": "nasa_temperature_avg",
    "temperature_max": "nasa_temperature_max",
    "temperature_min": "nasa_temperature_min",
    "humidity": "nasa_humidity",
    "wind_speed": "nasa_wind_speed",
    "precipitation": "nasa_precipitation"
})

# -----------------------------------------
# STEP 6: JOIN ALL THREE DATASETS
# Join using city and date
# Outer join helps identify missing records
# -----------------------------------------

integrated_df = pd.merge(
    air_daily,
    weather_daily,
    on=["city", "date"],
    how="outer"
)

integrated_df = pd.merge(
    integrated_df,
    nasa_daily,
    on=["city", "date"],
    how="outer"
)

# Sort final dataset
integrated_df = integrated_df.sort_values(
    ["city", "date"]
).reset_index(drop=True)

# -----------------------------------------
# STEP 7: VALIDATE INTEGRATED DATA
# -----------------------------------------

print("\n========== INTEGRATION SUMMARY ==========")

print("Air Daily Records:", len(air_daily))
print("Weather Daily Records:", len(weather_daily))
print("NASA Daily Records:", len(nasa_daily))

print("\nIntegrated Records:", len(integrated_df))
print("Unique Cities:", integrated_df["city"].nunique())
print("Unique Dates:", integrated_df["date"].nunique())

print("\nMissing Values:")
print(integrated_df.isnull().sum())

print("\nDuplicate City-Date Records:",
      integrated_df.duplicated(["city", "date"]).sum())

# -----------------------------------------
# STEP 8: DISPLAY FINAL DATASET
# -----------------------------------------

print("\n========== FINAL INTEGRATED DATASET ==========")

display(integrated_df.head(10))

print("\nDataset Shape:", integrated_df.shape)
print("\nDataset Columns:")
print(integrated_df.columns.tolist())
```

```python

import requests
import pandas as pd
from datetime import datetime

# Project cities and coordinates
cities = {
    "Delhi": (28.6139, 77.2090),
    "Mumbai": (19.0760, 72.8777),
    "Bengaluru": (12.9716, 77.5946),
    "Chennai": (13.0827, 80.2707),
    "Kolkata": (22.5726, 88.3639),
    "Hyderabad": (17.3850, 78.4867)
}

# API endpoint
url = "https://air-quality-api.open-meteo.com/v1/air-quality"

all_data = []

# Extract data for each city
for city, (latitude, longitude) in cities.items():

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "hourly": "pm10,pm2_5,nitrogen_dioxide,sulphur_dioxide,carbon_monoxide,ozone",
        "past_days": 1,
        "forecast_days": 1,
        "timezone": "Asia/Kolkata"
    }

    response = requests.get(url, params=params, timeout=30)

    # Check API response
    response.raise_for_status()
    data = response.json()

    # Convert API response into a DataFrame
    city_df = pd.DataFrame(data["hourly"])
    city_df["city"] = city
    city_df["latitude"] = latitude
    city_df["longitude"] = longitude

    all_data.append(city_df)

    print(f"{city}: API data extracted successfully")

# Combine all city data
air_quality_df = pd.concat(all_data, ignore_index=True)

print("Total extracted records:", len(air_quality_df))

display(air_quality_df)
```
